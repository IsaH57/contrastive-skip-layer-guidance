```python
import torch
import numpy as np
import matplotlib.pyplot as plt

def interpolate_nans(tensor):
    """Interpolate NaN values in a 1D tensor using linear interpolation."""
    result = tensor.clone()
    nan_mask = torch.isnan(tensor)
    
    if nan_mask.any():
        # Get indices
        indices = np.arange(len(tensor))
        
        # Valid (non-NaN) values
        valid_mask = ~nan_mask.numpy()
        valid_indices = indices[valid_mask]
        valid_values = tensor[valid_mask].numpy()
        
        # Interpolate at NaN positions
        nan_indices = indices[nan_mask.numpy()]
        interpolated = np.interp(nan_indices, valid_indices, valid_values)
        result[nan_mask] = torch.from_numpy(interpolated)
    
    return result
```
